# Prompt 1 — Pengolahan Data Geomatika

## R — Role
Anda adalah seorang ahli geomatika dan geofisika yang berpengalaman dalam pengolahan data koordinat, survei, dan pemetaan.

## T — Task
Analisis data koordinat dan elevasi hasil survei geomatika untuk mengetahui persebaran titik pengamatan serta perubahan elevasi di wilayah penelitian.

## C — Context
Data hasil survei lapangan berupa koordinat dan elevasi perlu diolah agar dapat digunakan untuk membuat peta persebaran titik dan memahami kondisi topografi wilayah penelitian.

## I — Input
Gunakan data berikut:

| Titik | X (m) | Y (m) | Elevasi (m) |
|---|---:|---:|---:|
| P1 | 500 | 1000 | 25 |
| P2 | 550 | 1050 | 28 |
| P3 | 600 | 1100 | 32 |
| P4 | 650 | 1080 | 30 |
| P5 | 700 | 1150 | 35 |

## K — Constraints
- Gunakan satuan meter.
- Jangan mengubah data asli.
- Jelaskan setiap tahapan pengolahan.
- Perhatikan hubungan antara koordinat dan posisi titik.
- Jangan membuat data tambahan yang tidak diberikan.
- Jika terdapat keterbatasan data, jelaskan keterbatasannya.

## O — Output
Berikan:
1. Tabel data yang telah diperiksa.
2. Plot persebaran titik berdasarkan koordinat X dan Y.
3. Analisis perubahan elevasi.
4. Identifikasi titik dengan elevasi tertinggi dan terendah.
5. Interpretasi kondisi topografi berdasarkan data.
6. Kesimpulan.

## Q — Quality Criteria
Hasil harus memiliki koordinat yang benar, grafik mudah dibaca, satuan konsisten, dan interpretasi sesuai dengan data.

## V — Verification
Periksa kembali koordinat setiap titik dan pastikan posisi titik pada grafik sesuai dengan data X dan Y. Periksa kembali nilai elevasi tertinggi dan terendah.

# Prompt 2 — Analisis Geostatistika Menggunakan MATLAB

## R — Role
Anda adalah seorang ahli geostatistika dan pemrograman MATLAB yang berpengalaman dalam mengolah dan menganalisis data geofisika.

## T — Task
Analisis data anomali gravitasi menggunakan MATLAB untuk mengetahui karakteristik statistik data dan pola persebarannya secara spasial.

## C — Context
Data geofisika yang diperoleh dari beberapa titik pengukuran memiliki nilai yang berbeda-beda. Geostatistika dapat digunakan untuk mengetahui karakteristik data dan hubungan antara nilai parameter geofisika dengan lokasi pengukurannya.

## I — Input
Gunakan data berikut:

| Titik | X | Y | Anomali Gravitasi (mGal) |
|---|---:|---:|---:|
| P1 | 0 | 0 | 12.5 |
| P2 | 10 | 0 | 14.2 |
| P3 | 20 | 0 | 16.1 |
| P4 | 0 | 10 | 11.8 |
| P5 | 10 | 10 | 15.0 |
| P6 | 20 | 10 | 17.3 |

## K — Constraints
- Gunakan MATLAB.
- Gunakan kode yang sederhana dan mudah dipahami.
- Jelaskan fungsi setiap bagian kode.
- Jangan mengubah data asli.
- Gunakan satuan mGal untuk anomali gravitasi.
- Jangan menggunakan metode yang terlalu kompleks tanpa menjelaskan konsep dasarnya.

## O — Output
Berikan:
1. Kode MATLAB.
2. Nilai minimum, maksimum, rata-rata, dan standar deviasi.
3. Grafik persebaran data.
4. Analisis pola nilai anomali secara spasial.
5. Jika memungkinkan, buat analisis variogram sederhana.
6. Interpretasi hasil.
7. Kesimpulan.

## Q — Quality Criteria
Kode harus dapat dijalankan di MATLAB, menggunakan sintaks yang benar, menghasilkan grafik yang jelas, dan setiap hasil perhitungan harus dapat dijelaskan secara geostatistika.

## V — Verification
Periksa kembali hasil statistik menggunakan perhitungan manual sederhana. Pastikan nilai minimum, maksimum, dan rata-rata sesuai dengan data input. Periksa juga apakah grafik menunjukkan posisi titik yang benar.

# Prompt 3 — Analisis Geospasial Menggunakan ArcGIS

## R — Role
Anda adalah seorang ahli Sistem Informasi Geografis (GIS) dan geofisika yang berpengalaman menggunakan ArcGIS untuk analisis spasial.

## T — Task
Buat analisis kerawanan longsor menggunakan metode Weighted Overlay pada ArcGIS berdasarkan beberapa parameter lingkungan.

## C — Context
Kerawanan longsor dipengaruhi oleh berbagai faktor, salah satunya kemiringan lereng, curah hujan, dan penggunaan lahan. Setiap parameter memiliki pengaruh yang berbeda sehingga perlu diberikan bobot sebelum digabungkan secara spasial.

## I — Input
Terdapat tiga parameter:

1. Kemiringan lereng dengan bobot 50%.
2. Curah hujan dengan bobot 30%.
3. Penggunaan lahan dengan bobot 20%.

Contoh klasifikasi skor:

### Kemiringan Lereng
- 0–8° = skor 1
- 8–15° = skor 2
- 15–25° = skor 3
- >25° = skor 4

### Curah Hujan
- Rendah = skor 1
- Sedang = skor 2
- Tinggi = skor 3
- Sangat tinggi = skor 4

### Penggunaan Lahan
- Hutan = skor 1
- Perkebunan = skor 2
- Pertanian = skor 3
- Permukiman/lahan terbuka = skor 4

## K — Constraints
- Gunakan ArcGIS.
- Pastikan seluruh layer memiliki sistem koordinat yang sesuai.
- Jelaskan proses preprocessing.
- Jelaskan alasan pemberian skor dan bobot.
- Gunakan Weighted Overlay.
- Jangan mengubah data asli.
- Jangan menentukan parameter tambahan tanpa menjelaskan alasannya.

## O — Output
Berikan:
1. Tahapan pengolahan data di ArcGIS.
2. Tabel klasifikasi skor.
3. Tabel bobot parameter.
4. Proses Weighted Overlay.
5. Peta kerawanan longsor.
6. Kelas kerawanan rendah, sedang, dan tinggi.
7. Interpretasi hasil.
8. Kesimpulan.

## Q — Quality Criteria
Hasil harus memiliki sistem koordinat yang konsisten, skor dan bobot yang jelas, proses analisis yang dapat diikuti, serta peta akhir yang mudah dibaca.

## V — Verification
Periksa kembali sistem koordinat, klasifikasi skor, dan total bobot. Pastikan total bobot adalah 100%. Jika tersedia data kejadian longsor, bandingkan lokasi longsor aktual dengan hasil peta kerawanan.

# Prompt 4 — Analisis Seismogram Menggunakan SeisGram2K

## R — Role
Anda adalah seorang seismolog yang berpengalaman dalam analisis seismogram dan penggunaan perangkat lunak SeisGram2K.

## T — Task
Identifikasi waktu tiba gelombang P dan S pada rekaman seismogram menggunakan SeisGram2K, kemudian hitung selisih waktu tiba antara kedua gelombang tersebut.

## C — Context
Gelombang P merupakan gelombang yang umumnya tiba lebih dahulu dibandingkan gelombang S. Perbedaan waktu tiba P dan S dapat digunakan dalam analisis seismologi, termasuk untuk memperkirakan parameter gempa jika dikombinasikan dengan data dari beberapa stasiun.

## I — Input
Contoh data:
- Waktu tiba gelombang P (Tp) = 10:15:20
- Waktu tiba gelombang S (Ts) = 10:15:32
- Data rekaman seismogram tersedia dalam SeisGram2K.

## K — Constraints
- Gunakan SeisGram2K untuk mengamati dan melakukan picking.
- Jangan menentukan titik P dan S tanpa melihat karakter sinyal.
- Jelaskan ciri-ciri gelombang P dan S.
- Gunakan satuan detik untuk perhitungan S-P.
- Jika terdapat ketidakpastian dalam picking, jelaskan penyebabnya.

## O — Output
Berikan:
1. Letak gelombang P pada seismogram.
2. Letak gelombang S pada seismogram.
3. Nilai Tp.
4. Nilai Ts.
5. Perhitungan S-P.
6. Interpretasi hasil.
7. Kesimpulan.

Gunakan rumus:

S-P = Ts - Tp

## Q — Quality Criteria
Picking gelombang P dan S harus berdasarkan perubahan karakter sinyal pada seismogram. Hasil perhitungan harus menggunakan satuan waktu yang benar.

## V — Verification
Periksa kembali bahwa gelombang P tiba sebelum gelombang S. Hitung ulang selisih waktu:

S-P = 10:15:32 - 10:15:20 = 12 detik

Pastikan hasil tersebut sesuai dengan waktu yang terlihat pada seismogram.

# Prompt 5 — Analisis Medan dan Gelombang

## R — Role
Anda adalah seorang ahli fisika dan geofisika yang memahami konsep medan vektor, kalkulus vektor, dan fenomena gelombang.

## T — Task
Tentukan divergensi dari suatu medan vektor dan jelaskan makna fisis dari hasil perhitungan tersebut dalam konteks fisika medan.

## C — Context
Dalam fisika medan dan geofisika, medan vektor dapat digunakan untuk menggambarkan berbagai besaran seperti medan gravitasi, medan listrik, dan medan magnet. Divergensi digunakan untuk mengetahui karakteristik sumber atau penyebaran suatu medan pada suatu ruang.

## I — Input
Diberikan medan vektor:

F(x,y,z) = 2x i + 3y j + 4z k

Tentukan:

∇ · F

## K — Constraints
- Gunakan konsep kalkulus vektor.
- Tunjukkan setiap tahapan perhitungan.
- Jelaskan rumus yang digunakan.
- Jelaskan arti fisik dari hasil.
- Jangan langsung memberikan jawaban akhir tanpa proses.
- Gunakan notasi matematika yang benar.

## O — Output
Berikan:
1. Persamaan medan vektor.
2. Rumus divergensi.
3. Substitusi komponen medan.
4. Hasil akhir.
5. Interpretasi fisis.
6. Hubungan dengan konsep medan dalam geofisika.
7. Kesimpulan.

## Q — Quality Criteria
Perhitungan harus benar secara matematis, notasi harus jelas, dan interpretasi fisis harus sesuai dengan konsep divergensi.

## V — Verification
Periksa kembali setiap turunan parsial.

∇ · F =
∂Fx/∂x + ∂Fy/∂y + ∂Fz/∂z

Kemudian pastikan hasil akhir sesuai dengan komponen medan yang diberikan.
